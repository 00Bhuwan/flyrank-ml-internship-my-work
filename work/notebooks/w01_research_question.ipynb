## 3. Quick look at the data (2-3 real numbers)

The starter CSV already contains the page-level performance signals I need for refresh scoring. The code below loads it and shows how many pages are declining, how much traffic the dataset contains, and how many pages have a recent drop in impressions.

```python
import pandas as pd
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent.parent
path = repo_root / 'data' / 'raw' / 'content_refresh_anonymized.csv'

print('reading from:', path)
df = pd.read_csv(path)

print('rows in starter dataset:', len(df))
print('unique content rows:', df['content_id'].nunique())
print('unique clients:', df['client_id'].nunique())
print('pages trending down:', (df['trend_direction'] == 'down').sum())
print('share trending down: {:.1%}'.format((df['trend_direction'] == 'down').mean()))
print('median impressions_last_30d:', int(df['impressions_last_30d'].median()))
print('share with fewer recent impressions than prior 30 days: {:.1%}'.format((df['impressions_prev_30d'] > df['impressions_last_30d']).mean()))
print('pages with ai_sessions_90d > 0:', (df['ai_sessions_90d'] > 0).sum())
print('\nSample declining pages:')
print(df.loc[(df['trend_direction'] == 'down') & (df['impressions_last_30d'] > 0), ['content_id', 'client_id', 'impressions_last_30d', 'clicks_last_30d', 'impressions_prev_30d', 'trend_pct']].head(3).to_string(index=False))
```
